## 3. Crear tres gráficos interactivos con Plotly

```python
# Grafico de barras de la cantidad de lluvia por departamento

lluvia_mm = tabla.sort_values("lluvia_total_mm", ascending=False).head(10) # Ordeno la tabla por cantidad de lluvia y me quedo con los 10 primeros

graf_1 = px.bar(lluvia_mm, x="departamento", y="lluvia_total_mm",
                    title="Los 10 departamentos con mayor cantidad de lluvia en mm (Enero 2025 - Mayo 2025)",
                    labels={"departamento": "Departamento", "lluvia_total_mm": "Cantidad de lluvia (mm)"},
                    color="lluvia_total_mm", color_continuous_scale="teal")

graf_1.update_layout(coloraxis_showscale=False)

graf_1.show()
```

En el gráfico se puede observar que los 10 departamentos con mayor cantidad de lluvia acumulada en mm, durante los meses de enero a mayo del 2025, fueron Loreto, Madre de Dios, Áncash, Junín, Ucayali, Puno, San Martín, Huancavelica, Cajamarca y Pasco. 

```python
# Grafico de dispersión de la relación entre la cantidad de lluvia y las declaratorias según departamento

decla_lluvia =px.scatter(tabla[tabla["declaratorias"] > 0],
                x="lluvia_total_mm", y="declaratorias",
                hover_name="departamento",
                title="Relación entre cantidad de lluvia y declaratorias según departamento (Enero 2025 - Mayo 2025)",
                labels={"lluvia_total_mm": "Cantidad de lluvia (mm)", "declaratorias": "Cantidad de declaratorias"})

decla_lluvia.show()
```

Con el gráfico de dispersión se puede concluir que no existe una relación clara entre la cantidad de lluvia en un departamento y el número de declaratorias de emergencia emitidos, ya que los puntos están dispersos sin formar una tendencia. 

```python
# Histograma del número de departamentos según la cantidad de lluvia

hist_lluvia = px.histogram(tabla, x="lluvia_total_mm",
                nbins=10,
                color_discrete_sequence=["teal"],
                title="Número de departamentos según la cantidad de lluvia en mm (Enero 2025 - Mayo 2025)",
                labels={"lluvia_total_mm": "Cantidad de lluvia (mm)"})

hist_lluvia.update_layout(yaxis_title="Cantidad de departamentos", bargap=0.05)

hist_lluvia.show()
```

A partir del histograma, se precisa que la mayoria de departamentos (10) acumuló menos de 500 mm de lluvia, mientras solo dos departamentos presentaron más de 1500 mm de lluvia. 

## 4. Verificación rápida

```python
# Gráfico 1: Los 10 departamentos con mayor cantidad de lluvia en mm (Enero 2025 - Mayo 2025)

tabla.sort_values("lluvia_total_mm", ascending=False)[["departamento", "lluvia_total_mm"]].head(10)
```

Se puede comprobar que tanto la tabla como el gráfico muestran los mismos departamentos. 

```python
# Gráfico 2: Cantidad de lluvia por declaratorias según departamento (Enero 2025 - Mayo 2025)

# Mostrar los 5 departamentos con mayor cantidad de lluvia
tabla.sort_values("lluvia_total_mm", ascending=False)[["departamento", "lluvia_total_mm", "declaratorias"]].head(5)
```

```python
# Mostrar los 5 departamentos con mayor cantidad de declaratorias       
tabla.sort_values("declaratorias", ascending=False)[["departamento", "lluvia_total_mm", "declaratorias"]].head(5)
```

```python
Se puede observar que los departamentos con mayor cantidad de lluvia no necesariamente son los mismos que tienen mayor cantidad de declaratorias, lo que indica que la relación entre la cantidad de lluvia y las declaratorias no es directa y puede depender de otros factores.
```

```python
# Grafico 3: Número de departamentos según la cantidad de lluvia en mm (Enero 2025 - Mayo 2025)

pd.cut(tabla["lluvia_total_mm"], bins=[0, 500, 1000, 1500, 2000, 2500], include_lowest=True).value_counts().sort_index()
```

Se comprueba que 10 departamentos, es decir, la mayoría presenta hasta 500 mm de lluvia acumulada, mientras solo 2 muestran más de 1500 mm de lluvia acumulada para el periodo analizado.
