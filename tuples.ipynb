# Assingment 1 - Tuples 

```python
ages = (20, 25, 22, 30, 28)
```

```python
# 1. Print the tuple

print (ages)
```

```python
# 2.1 Display the first element
print (ages[0])
```

```python
# 2.2 Display the last element 
print (ages[-1])
```

```python
# 3.2 Display the minimum value

print (min(ages))
```

```python
# 3.3. Display the number of elements

print (len(ages))
```
