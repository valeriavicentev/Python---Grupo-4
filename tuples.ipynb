# Assingment 1 - Tuples 

```python
ages = (20, 25, 22, 30, 28)
```

```python
# 1. Print the tuple

print (ages)
```

```python
# 2.1 Display the first element
print (ages[0])
```

```python
# 2.2 Display the last element 
print (ages[-1])
```

```python
# 3.1 Display the maximum value

print(max(ages))
```

```python
# 3.2 Display the minimum value

print (min(ages))
```

```python
# 3.3. Display the number of elements

print (len(ages))
```

```python
# 4. Try to modify one of the elements of the tuple

ages[0]=32
```

# 5. Write a brief comment in the script explaining why Python does not allow this modification

Python does not allow this modification because the object "tuples" is an item container that cannot be changed. We can display certain items of it, however, we cannot change them, remove them, or add new ones once the tuple has been defined.
