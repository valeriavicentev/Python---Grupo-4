```python
# 1. Lista original de calificaciones

grades = [15, 18, 12, 20, 16]
print("Lista original:", grades)
```

```python
# 2. Agregar una nueva calificación a la lista
grades.append(17)

# 3. Ordenar la lista de calificaciones de menor a mayor
grades.sort()

# 4. Calcular la calificación más alta, la más baja y el número total de calificaciones
calificacion_mas_alta = max(grades)
calificacion_mas_baja = min(grades)
total_calificaciones = len(grades)

print("Calificación más alta:", calificacion_mas_alta)
print("Calificación más baja:", calificacion_mas_baja)
print("Número total de calificaciones:", total_calificaciones)

# 5. Imprime la lista final
print("Lista final ordenada:", grades)
```
